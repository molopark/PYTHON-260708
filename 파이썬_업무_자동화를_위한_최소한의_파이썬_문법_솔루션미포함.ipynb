````python
def calculate_bmi(weight, height):
    # 키를 미터 단위로 변환
    height_in_meters = height / 100

    # BMI 계산
    bmi = weight / (height_in_meters ** 2)

    return bmi
weight = float(input("몸무게(kg)를 입력하세요: "))
height = float(input("키(cm)를 입력하세요: "))

bmi = calculate_bmi(weight, height)

print("당신의 BMI는 {:.2f}입니다.".format(bmi))
````

## 쥬피터노트북 사용법

주피터 노트북 사용 방법

- 주피터 노트북 기본 경로
    - 노트북 파일이 저장되는 위치
    - `C:\Users\<사용자계정>`
    - 경로설정을 윈도우키 > jupyter > 오른쪽클릭 > 파일위치열기 > jupyter notebook > 오른쪽클릭 > 속성 > %USERNAME을 원하는 경로로 변경


- 노트북 생성
    - 우측 상단 `New` 버튼 클릭
    - `Python 3` 클릭

- 노트북 제목 변경
    - 좌측 상단 `Untitled` 클릭
    - 변경할 제목 입력 후 `Rename` 버튼 클릭


- 설명 셀 입력
    - 메뉴 바 중앙 드랍다운 메뉴 `Markdown` 선택
    - 입력 창 왼쪽에 아무 것도 없으면 설명 셀 (파란색)
    - 셀 클릭 후 내용 입력
    - 제목은 `#` 을 붙여 입력
    - `#` 이 늘어나면 하위 제목

- 실행 및 코드 셀 입력
    - Run 버튼을 클릭하면 현재 셀이 실행되고 다음 셀이 생성됨
    - 입력 창 왼쪽에 `In [ ]:` 이 있으면 코드 셀 (초록색)
    - 코드 입력



- 실행 순번 및 결과 확인
    - Run 버튼을 클릭하면 코드 셀이 실행되어 왼쪽에 실행 순번이 매겨지고 아래에 실행 결과가 출력됨
    - 코드 셀은 실행 상태가 순번대로 연결됨 (셀 위치가 바뀌어도 실행 순서는 순번을 따름)


- 셀 관리
    - 셀 추가, 셀 잘라내기, 셀 복사하기, 셀 붙여넣기, 현재 셀을 위로 밀기, 현재 셀을 아래로 밀기 등 여러 기능

- 주요 단축키
    - `[Esc]` : 셀 선택 모드 (또는 `[Ctrl]+[m]`)
        - `[m]` : 설명 셀로 변경
        - `[y]` : 코드 셀로 변경
        - `[a]` : 현재 셀 위에 셀 추가
        - `[b]` : 현재 셀 아래에 셀 추가
        - `[x]` : 현재 셀 잘라내기
        - `[c]` : 현재 셀 복사하기
        - `[v]` : 현재 셀 아래에 붙여넣기
        - `[d][d]` : 현재 셀 삭제
    - `[Enter]` : 셀 입력 모드
        - `[Ctrl]+[Enter]` : 현재 셀 실행
        - `[Shift]+[Enter]` : 현재 셀 실행 후 다음 셀 생성

- `[Ctrl]+[s]` : 저장
- `[Ctrl]+[z]` : 저장
- `[Ctrl]+[shift]+[z]` : 저장

### D-1.데이터구조란?

엑셀을 예로 들어보면, 행/열로 우리가 원하는 데이터를 저장하고 우리가 원할때 그 데이터를 가져올 수 있는 데이터구조

우리가 알고 있는 표와 같다.

그런데, 코딩에서는 단순 2차원인 표뿐만 아니라 3차,4차 N차원까지 자료를 저장할 수 있다.

그래서 데이터 구조를 이해하면 새로운 환경으로 진입할 수 있다.

#### D-1-1.데이터 구조의 종류

````python
import time

def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Execution time: {end_time - start_time} seconds")
        return result
    return wrapper

@timer_decorator
def example_function(duration):
    time.sleep(duration)

example_function(2)  # 출력: Execution time: 2.002 seconds
````

## 클래스란?
클래스는 객체를 표현하기 위한 문법. 클래스에는 크게 속성과 메소드 두가지로 구분되는데, 당장은 속성의 경우 매개변수를 받고 사용하기 위한 값을 정의하며 메소드는 만들어진 속성들을 이용해 어떤 행위를 하는 실행 코드

EX) 군대게임의 클래스(직업)를
각각의 클래스(육군, 공군, 해군 등)가 있고
그 안의 속성(힘, 민첩, 지력, 운, 원거리공격력, 근접공격력)이 있으며 각각의 메소드(총발사, 수류탄던지기, 대포쏘기 등)가 있다고 생각할 수 있음

하지만, 업무자동화에서 클래스를 널리 사용할 일은 그리 많지 않음. 향후 웹개발(파이썬기반 웹프레임워크 장고) 등을 활용할 때 본격적으로 사용하게 됨

````python
result1 = 0
result2 = 0
result3 = 0
result4 = 0

def add1(num):
    global result1
    result1 += num
    return result1

def add2(num):
    global result2
    result2 += num
    return result2

def minus1(num):
    global result3
    result3 = num - result3
    return result3

def minus2(num):
    global result4
    result4 = num - result4
    return result4

print(add1(3))
print(add1(4))
print(add2(3))
print(add2(7))
print(minus1(3))
print(minus1(7))
print(minus2(3))
print(minus2(7))
````

````python
class Calculator:
    def __init__(self):
        self.result = 0

    def add(self, num):
        self.result += num
        return self.result

    def minus(self, num):
        self.result = self.result-num
        return self.result

cal1 = Calculator()
cal2 = Calculator()

print(cal1.add(3))
print(cal1.add(4))
print(cal2.add(3))
print(cal2.add(7))

print(cal1.minus(3))
print(cal1.minus(4))
print(cal2.minus(3))
print(cal2.minus(7))
````

````python
%%writefile main.cpp
#include<iostream>
#include<vector>
#include<cmath>
#include<chrono>

bool is_prime(int num) {
    if (num < 2) {
        return false;
    }
    for (int i = 2; i <= int(sqrt(num)); ++i) {
        if (num % i == 0) {
            return false;
        }
    }
    return true;
}

int main() {
    auto start_time = std::chrono::high_resolution_clock::now();

    std::vector<int> primes;
    for (int number = 2; number <= 10000; ++number) {
        if (is_prime(number)) {
            primes.push_back(number);
        }
    }

    auto end_time = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double> elapsed_time = end_time - start_time;

    std::cout << "소수를 찾는 데에 걸린 시간: " << elapsed_time.count() << "초" << std::endl;
    std::cout << "10,000 이하의 소수: ";
    for (const auto& prime : primes) {
        std::cout << prime << " ";
    }
    std::cout << std::endl;

    return 0;
}
````
